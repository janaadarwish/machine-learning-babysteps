```python
DimCustomer = DimCustomer.drop(columns=['Valid From', 'Valid To'])
print(DimCustomer.columns.tolist())
```

```python
print(DimCustomer['Credit Limit'].unique())
```

```python
# Step 1: remove the '? ' prefix
DimCustomer['Credit Limit'] = DimCustomer['Credit Limit'].str.replace('? ', '', regex=False)

# Step 2: remove commas
DimCustomer['Credit Limit'] = DimCustomer['Credit Limit'].str.replace(',', '', regex=False)

# Step 3: replace the leftover '-' with NaN (missing value)
DimCustomer['Credit Limit'] = DimCustomer['Credit Limit'].replace('-', pd.NA)

# Step 4: convert the column to an actual number
DimCustomer['Credit Limit'] = pd.to_numeric(DimCustomer['Credit Limit'])

# Check the result
print(DimCustomer['Credit Limit'].dtype)
print(DimCustomer['Credit Limit'].describe())
print(DimCustomer['Credit Limit'].isnull().sum())
```

```python
DimStockItem['Recommended Retail Price'] = DimStockItem['Recommended Retail Price'].str.replace('? ', '', regex=False)
DimStockItem['Recommended Retail Price'] = DimStockItem['Recommended Retail Price'].str.replace(',', '', regex=False)
DimStockItem['Recommended Retail Price'] = DimStockItem['Recommended Retail Price'].replace('-', pd.NA)
DimStockItem['Recommended Retail Price'] = pd.to_numeric(DimStockItem['Recommended Retail Price'])

print(DimStockItem['Recommended Retail Price'].dtype)
print(DimStockItem['Recommended Retail Price'].describe())
```
