```python
print(
    DimStockItem['Recommended Retail Price'].unique()[:10]
)
print(DimStockItem['Valid From'].unique()[:5])
print(DimStockItem['Valid To'].unique()[:5])
```

```python
print(DimStockItem['Valid To'].value_counts().head(20))
print(DimStockItem['Valid From'].nunique())
print(DimStockItem['Valid From'].value_counts().head(10))
```

```python
DimStockItem = DimStockItem.drop(
    columns=['Valid From',
             'Valid To',
             'Photo',
             'Discount'])
print(DimStockItem.columns.tolist())
```

```python
DimStockItem['Recommended Retail Price'] = DimStockItem['Recommended Retail Price'].str.replace('? ', '', regex=False)
DimStockItem['Recommended Retail Price'] = DimStockItem['Recommended Retail Price'].str.replace(',', '', regex=False)
DimStockItem['Recommended Retail Price'] = DimStockItem['Recommended Retail Price'].replace('-', pd.NA)
DimStockItem['Recommended Retail Price'] = pd.to_numeric(DimStockItem['Recommended Retail Price'])

print(DimStockItem['Recommended Retail Price'].dtype)
print(DimStockItem['Recommended Retail Price'].describe())
```
